```
N = 44
```

```
i = arange(0, N, 1)
```

```
x = index_build(i, lambda i: 4 * (i / N) + 6 * (i / N)**2)
disp(x[i])
```

```
def q(t):
    if t <= 5:
        return 1 + 0.2 * t
    elif t <= 8:
        return 2 + (t - 5)**2
    return 11 - 1.5 * (t - 8)
q = elementwise(q)
```
